# 下载西瓜语义分割数据集

## 1. 环境初始化与工作目录锁定

本节负责将工作路径强行锁定到 `mmsegmentation` 主目录，防止后续文件下载和解压路径错乱。

```python
import os

# 锁定在 AutoDL 上的绝对路径
WORK_DIR = '/root/LearningMMSegmentation/mmsegmentation'

if os.path.exists(WORK_DIR):
    os.chdir(WORK_DIR)
    print(f"✅ 成功进入工作目录: {os.getcwd()}")
else:
    print(f"❌ 错误：找不到路径 {WORK_DIR}，请检查你的文件夹结构。")
```

## 2. 下载并解压西瓜语义分割数据集

如果下载速度过慢，请在 AutoDL 终端执行 **source /etc/network_turbo** 开启学术加速。
本步骤会自动清理可能残留的旧文件夹，确保每次运行都能得到一份纯净的数据。

```python
# 1. 删除原有的旧数据集（防干扰）
!rm -rf data/Watermelon87_Semantic_Seg_Mask.zip data/Watermelon87_Semantic_Seg_Mask

# 2. 下载压缩包到 data/ 目录
!wget https://zihao-openmmlab.obs.cn-east-3.myhuaweicloud.com/20230130-mmseg/dataset/watermelon/Watermelon87_Semantic_Seg_Mask.zip -P data

# 3. 安静模式解压 (-q) 到 data/ 目录，避免满屏输出干扰视线
!unzip -q data/Watermelon87_Semantic_Seg_Mask.zip -d data/

# 4. 删除压缩包释放硬盘空间
!rm data/Watermelon87_Semantic_Seg_Mask.zip

print("✅ 西瓜数据集下载并解压完成，存放于 data/ 目录下。")
```

## 3. 下载测试素材（带防重复机制）

下载用于后续模型推理测试的图像与视频素材。
使用 **-O** 参数强制覆盖同名文件，防止因多次运行产生 **xxx.jpg.1** 的冗余文件。

```python
# 确保 data 目录存在
!mkdir -p data

print("⏳ 开始下载测试素材...")

# 强制命名并覆盖保存至 data/ 目录
!wget -q https://zihao-openmmlab.obs.cn-east-3.myhuaweicloud.com/20230130-mmseg/watermelon/data/watermelon_test1.jpg -O data/watermelon_test1.jpg
!wget -q https://zihao-openmmlab.obs.cn-east-3.myhuaweicloud.com/20230130-mmseg/watermelon/data/video_watermelon_2.mp4 -O data/video_watermelon_2.mp4
!wget -q https://zihao-openmmlab.obs.cn-east-3.myhuaweicloud.com/20230130-mmseg/watermelon/data/video_watermelon_3.mov -O data/video_watermelon_3.mov

print("✅ 所有测试素材已成功覆盖保存至 data/ 目录下！")
```

## 4. 一键清理系统垃圾文件

在解压文件或使用 Jupyter 时，系统经常会偷偷生成 **.DS_Store (Mac特产)**, **__MACOSX** 和 **.ipynb_checkpoints** 等隐藏缓存。如果不清理，后续用代码遍历文件夹时极易引发报错。

```python
# 🌟 重构亮点：使用极简的单行 Shell 命令，一键全自动查找并强制删除所有垃圾文件
!find . -iname '__MACOSX' -exec rm -rf {} +
!find . -iname '.DS_Store' -exec rm -rf {} +
!find . -iname '.ipynb_checkpoints' -exec rm -rf {} +

print("✅ 系统隐藏文件与冗余缓存已全部清理完毕！")
```
