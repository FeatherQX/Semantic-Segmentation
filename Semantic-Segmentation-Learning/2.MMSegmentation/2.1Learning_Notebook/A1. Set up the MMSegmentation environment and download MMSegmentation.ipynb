# 配置MMSegmentation相关环境并下载MMSegmentation

在 AutoDL 网页版的 JupyterLab 界面，点击左上角的 File -> New -> Terminal，打开终端窗口。

## 1.初始化 Conda

AutoDL 默认使用的 Shell 是 bash 。为使得其认识 conda activate 这个命令，需要先“初始化”。

```python
conda init bash
```

## 2.创建虚拟环境

输入以下命令创建一个专属的虚拟环境。给它起名叫"mmseg_env"（你也可以叫其他名字）。指定 Python 版本为 3.7（！！！初学者请配置相同版本的Python！！！）：

```python
conda create -n mmseg_env python=3.7 -y
```

## 3.激活环境"mmseg_env"

```python
conda activate mmseg_env
```

## 4.在环境"mmseg_env"中，配置相关依赖包，依次输入以下代码

安装 PyTorch

```python
pip install torch==1.10.1+cu113 torchvision==0.11.2+cu113 torchaudio==0.10.1+cu113 -f https://download.pytorch.org/whl/cu113/torch_stable.html
```

安装 OpenMMLab 专属包管理工具 (MIM)

```python
pip install -U openmim
```

安装深度学习基础训练引擎 (MMEngine)

```python
mim install mmengine
```

安装计算机视觉基础算法库 (MMCV)

```python
mim install mmcv==2.0.0
```

安装 MMSegmentation 源码包

```python
pip install "mmsegmentation==1.1.2"
```

安装 Jupyter 进度条显示插件

```python
pip install ipywidgets
```

## 5.让 Jupyter Notebook 识别新环境"mmseg_env"

安装 IPython 内核通信包

```python
conda install ipykernel -y
```

将虚拟环境注册到 Jupyter 菜单

```python
python -m ipykernel install --user --name seg_env --display-name "Python (mmseg_env)"
```

# 配置完成！！！
